# Literature Review: Appliance Energy Prediction Using Multiple Linear Regression

## 1. Introduction

Residential energy consumption is influenced by household activities, indoor environmental conditions, outdoor weather, and the operation of electrical appliances. Understanding these relationships is important for energy monitoring, consumption forecasting, building energy management, and the development of data-driven energy optimization systems.

Machine learning provides a statistical approach to analyzing household energy consumption by learning relationships between measured environmental conditions and observed energy use. Among the available techniques, Linear Regression offers an interpretable baseline for examining how multiple explanatory variables are associated with a continuous energy consumption target. Unlike more complex predictive models, it provides coefficients that can be examined to understand the direction and magnitude of conditional relationships between predictors and the target, subject to assumptions about the data and model specification.

This project uses the *Appliances Energy Prediction* dataset, collected in a low-energy residential building in Belgium. The dataset combines appliance energy consumption, lighting energy, indoor temperature and humidity measurements, and outdoor weather observations. Its combination of environmental and energy measurements makes it suitable for investigating whether sensor and weather variables can explain or predict appliance energy consumption.

The primary objective is to develop a Multiple Linear Regression model that estimates appliance energy consumption from selected environmental and, where appropriate, temporal features. The project will follow a reproducible machine learning workflow encompassing exploratory data analysis (EDA), data preparation, feature selection, model development, evaluation, and deployment.

## 2. Review of Foundational Research

The principal reference for this project is the study by Candanedo, Feldheim, and Deramaix (2017), titled *Data driven prediction models of energy use of appliances in a low-energy house*, published in *Energy and Buildings*.

The authors investigated data-driven approaches to predicting appliance energy consumption using measurements from indoor wireless sensors, lighting energy consumption, and external weather observations. The study examined the predictive value of different environmental variables and compared machine learning approaches for estimating appliance energy use.

Several findings from this research are particularly relevant to the present project.

First, the study reported that incorporating weather-station measurements improved appliance energy prediction. This suggests that external environmental conditions may provide useful information beyond indoor sensor measurements alone.

Second, outdoor variables, including temperature, atmospheric pressure, and wind speed, were identified as important parameters in the reported prediction experiments. These findings motivate an investigation of whether outdoor weather variables improve the predictive performance of our Multiple Linear Regression model.

Third, measurements from indoor sensor locations, including the kitchen, laundry room, and living room, ranked highly in importance among the wireless sensor network variables. This provides a research-based reason to examine the relationships between appliance energy consumption and measurements from different rooms.

These findings establish a useful starting point for our exploratory analysis. However, they should not be treated as guaranteed outcomes for our fitted model. Feature importance depends on the model, evaluation procedure, dataset characteristics, and the relationships among predictors.

The original research also highlights the value of evaluating several potential predictors rather than assuming that every available variable contributes equally to prediction. Building on this principle, the present project will investigate environmental and temporal relationships, assess multicollinearity, and compare alternative feature sets using an appropriate evaluation strategy.

## 3. Dataset Description and Data Collection

The dataset is publicly available through the UCI Machine Learning Repository under the title *Appliances Energy Prediction*. It is a multivariate time-series regression dataset containing 19,735 observations collected at 10-minute intervals over approximately 4.5 months.

The measurements were collected in a low-energy residential building in Belgium. Indoor temperature and humidity measurements were obtained using a ZigBee wireless sensor network. Appliance energy consumption was measured using energy meters, while lighting energy consumption was recorded separately. External weather observations from the weather station associated with Chièvres Airport were incorporated into the dataset using timestamps.

Each observation represents a measurement interval and contains appliance energy consumption, lighting energy, environmental sensor measurements, and weather information. The resulting dataset allows us to investigate relationships between household energy use and the environmental conditions recorded during the observation period.

The dataset contains 29 columns: one target variable, `Appliances`, and 28 potential input features. These include a timestamp, lighting energy consumption, nine temperature measurements, nine relative humidity measurements, six outdoor weather variables, and two random variables.

Although the dataset is suitable for regression analysis, its structure introduces important methodological considerations. The observations are temporally ordered and may exhibit autocorrelation, recurring household routines, and changing environmental conditions. Consequently, randomly splitting observations into training and test sets may produce an overly optimistic estimate of predictive performance. A chronological evaluation strategy will therefore be considered to assess whether the model generalizes to later observations.

Furthermore, because the data originate from one residential building over a limited period, the model's performance cannot automatically be generalized to other households, buildings, climates, or seasons.

## 4. Description of Variables

### 4.1 Target Variable: Appliances

The primary target variable is `Appliances`, which represents the recorded energy consumption of household appliances in watt-hours (Wh).

The objective is to estimate appliance energy consumption using a selected set of indoor environmental measurements, outdoor weather variables, and potentially engineered temporal features.

This is a continuous numerical target, making regression an appropriate modeling framework.

The target should not be confused with total household electricity consumption. Lighting energy is represented separately by the `lights` variable. The model will therefore estimate appliance energy consumption rather than total residential electricity demand.

### 4.2 Timestamp Variable

The `date` variable records the date and time associated with each observation.

Time may be relevant because appliance usage can vary with daily household routines, activity periods, and changes in environmental conditions. The timestamp can be used to investigate temporal patterns and construct features such as hour of day and day of week.

However, the raw timestamp should not automatically be included as an ordinary numerical predictor. Instead, it will be parsed and analyzed to determine which temporal representations, if any, improve predictive performance.

The timestamp will also be used to preserve chronological order when constructing training and test datasets.

### 4.3 Lighting Energy Consumption

The `lights` variable records energy consumption associated with lighting, measured in watt-hours.

Lighting and appliance consumption may respond to similar household activities or occupancy patterns. For this reason, lighting energy may contain information associated with the target.

Nevertheless, its inclusion requires consideration of the intended prediction scenario. If the model is intended to predict appliance energy consumption before a measurement interval begins, lighting consumption measured during that same interval may not be available at prediction time. Using it could create an unrealistic prediction setup.

The project will therefore distinguish between contemporaneous estimation, in which measurements from the same interval are available, and advance forecasting, in which only information available before the target interval may be used.

### 4.4 Indoor Temperature Variables

The temperature variables are:

- `T1`: Kitchen temperature.
- `T2`: Living room temperature.
- `T3`: Laundry room temperature.
- `T4`: Office temperature.
- `T5`: Bathroom temperature.
- `T6`: Temperature outside the building on the north side.
- `T7`: Ironing room temperature.
- `T8`: Teenager's room temperature.
- `T9`: Parents' room temperature.

These variables represent indoor or near-building temperature measurements and may capture environmental changes associated with cooking, laundry, heating, cooling, ventilation, and other household activities.

The relationship between temperature and appliance energy consumption may not be strictly linear. Additionally, temperatures measured in different rooms may be strongly correlated because they respond to common environmental conditions.

Consequently, the analysis will examine distributions, correlations, and relationships with the target before determining which temperature variables should be retained.

### 4.5 Indoor Relative Humidity Variables

The relative humidity variables are:

- `RH_1`: Kitchen relative humidity.
- `RH_2`: Living room relative humidity.
- `RH_3`: Laundry room relative humidity.
- `RH_4`: Office relative humidity.
- `RH_5`: Bathroom relative humidity.
- `RH_6`: Relative humidity outside the building on the north side.
- `RH_7`: Ironing room relative humidity.
- `RH_8`: Teenager's room relative humidity.
- `RH_9`: Parents' room relative humidity.

These measurements represent relative humidity as a percentage.

Humidity may be associated with household activities, ventilation, outdoor weather, and changes in indoor environmental conditions. For example, moisture-producing activities may coincide with changes in humidity and appliance use.

However, the strength and direction of these relationships must be established empirically. Correlation alone does not demonstrate a causal relationship, and the presence of correlated humidity measurements may complicate the interpretation of individual regression coefficients.

### 4.6 Outdoor Weather Variables

The dataset contains six outdoor weather variables:

- `T_out`: Outdoor air temperature, measured in degrees Celsius.
- `Press_mm_hg`: Atmospheric pressure, measured in millimetres of mercury.
- `RH_out`: Outdoor relative humidity, expressed as a percentage.
- `Windspeed`: Wind speed, measured in metres per second.
- `Visibility`: Visibility distance, measured in kilometres.
- `Tdewpoint`: Dew-point temperature, measured in degrees Celsius.

These variables describe external environmental conditions that may influence indoor temperatures, ventilation, and household energy use.

The original study reported predictive value from weather-station measurements, making these variables particularly relevant to the present investigation.

Their contribution will be assessed by comparing models that use indoor environmental measurements with models that additionally incorporate outdoor weather variables.

Potential correlations among outdoor temperature, dew-point temperature, and relative humidity will also be examined to determine whether multicollinearity affects the stability of regression coefficients.

### 4.7 Random Variables

The dataset contains two additional variables, `rv1` and `rv2`.

These variables were included by the dataset creators as random controls for evaluating predictive models and non-predictive attributes. They do not represent meaningful physical measurements of household environmental conditions.

They will therefore be excluded from the primary model. Their presence will nevertheless be documented during data inspection to demonstrate deliberate feature selection rather than indiscriminate use of all available columns.

## 5. Target Definition and Regression Framework

The primary modeling objective is to predict `Appliances`, measured in watt-hours, using selected environmental and temporal features.

Multiple Linear Regression represents the target as a linear combination of explanatory variables:

\[
\widehat{y}_t =
\beta_0 +
\beta_1X_{1,t} +
\beta_2X_{2,t} +
\cdots +
\beta_pX_{p,t}
\]

where:

- \(\widehat{y}_t\) is the predicted appliance energy consumption for observation \(t\).
- \(X_{1,t}, X_{2,t}, \ldots, X_{p,t}\) are the selected predictor variables.
- \(\beta_0\) is the intercept.
- \(\beta_1, \beta_2, \ldots, \beta_p\) are regression coefficients estimated from training data.
- \(p\) is the number of predictors included in the model.

The model will initially be developed as an explanatory and predictive model for contemporaneous appliance energy consumption, subject to verification of feature availability. A separate forecasting interpretation may be explored if the project later incorporates lagged measurements and strictly time-appropriate predictors.

The coefficients will be interpreted as conditional linear associations between individual predictors and the target, holding the other included predictors constant. They will not be interpreted as causal effects.

The primary modeling objective is to determine whether environmental and temporal variables provide useful information for estimating appliance energy consumption and whether the resulting model performs adequately on observations not used for training.

## 6. Research Hypotheses

The literature and the physical context of the dataset motivate five hypotheses that will guide the exploratory data analysis and subsequent modeling.

These hypotheses are provisional research propositions. Their validity will be assessed using the observed data, suitable statistical diagnostics, and out-of-sample model evaluation.

### Hypothesis 1: Indoor Environmental Conditions and Appliance Energy Consumption

**Research hypothesis:** Indoor temperature and relative humidity measurements are associated with appliance energy consumption.

**Null hypothesis (\(H_0\)):** The selected indoor environmental predictors have no linear association with appliance consumption after accounting for the other predictors in the specified model.

**Alternative hypothesis (\(H_1\)):** At least one indoor environmental predictor has a non-zero conditional linear association with appliance consumption.

**Rationale:** Indoor environmental measurements may reflect conditions associated with household activities and the operation of heating, cooling, ventilation, and other appliances. The original study also reported that measurements from several indoor sensor locations were important in its prediction experiments.

**EDA approach:** Examine the distributions of temperature and humidity variables, construct scatterplots against `Appliances`, and calculate correlations. Particular attention will be given to measurements from the kitchen, laundry room, and living room.

**Evaluation approach:** Fit a regression model with selected indoor environmental predictors and examine coefficient estimates, uncertainty, residual diagnostics, and out-of-sample performance.

A statistically significant coefficient will not, by itself, establish that a predictor materially improves prediction or that it causes changes in energy consumption.

### Hypothesis 2: Additional Predictive Value of Outdoor Weather Variables

**Research hypothesis:** Outdoor weather variables provide useful predictive information about appliance energy consumption beyond indoor environmental measurements alone.

**Null hypothesis (\(H_0\)):** Adding outdoor weather variables does not improve out-of-sample predictive performance.

**Alternative hypothesis (\(H_1\)):** Adding outdoor weather variables improves out-of-sample predictive performance.

**Rationale:** The original study reported improved prediction when weather-station measurements were incorporated. Outdoor weather may help characterize environmental conditions that influence indoor temperatures, ventilation, and household behavior.

**EDA approach:** Examine the distributions and relationships of `T_out`, `Press_mm_hg`, `RH_out`, `Windspeed`, `Visibility`, and `Tdewpoint`. Investigate their associations with the target and their correlations with indoor sensor variables.

**Evaluation approach:** Compare two otherwise comparable regression models using the same chronological training and test periods. The baseline model will use selected indoor environmental predictors, while the expanded model will also include outdoor weather variables.

The comparison will use metrics such as mean absolute error (MAE), root mean squared error (RMSE), and coefficient of determination (\(R^2\)).

### Hypothesis 3: Temporal Patterns in Appliance Energy Consumption

**Research hypothesis:** Appliance energy consumption varies systematically with time of day and day of week.

**Null hypothesis (\(H_0\)):** The inclusion of engineered temporal features does not improve out-of-sample predictive performance.

**Alternative hypothesis (\(H_1\)):** At least one engineered temporal feature improves out-of-sample predictive performance.

**Rationale:** Household appliance usage may follow recurring daily routines and weekly activity patterns. Since the dataset contains observations at 10-minute intervals, it allows temporal patterns to be investigated.

**EDA approach:** Parse the `date` variable and examine appliance energy consumption over time. Compare average consumption across hours and weekdays, and investigate whether the target exhibits recurring patterns, abrupt changes, or periods of elevated consumption.

**Evaluation approach:** Compare a model without engineered temporal features against one incorporating suitable time variables, such as hour of day and day of week. Any derived features will be constructed without using future information.

A chronological test set will be retained to assess whether the observed patterns provide predictive value beyond the training period.

### Hypothesis 4: Association Between Lighting and Appliance Energy Consumption

**Research hypothesis:** Lighting energy consumption is positively associated with appliance energy consumption.

**Null hypothesis (\(H_0\)):** The conditional linear coefficient of `lights` is zero.

**Alternative hypothesis (\(H_1\)):** The conditional linear coefficient of `lights` is positive.

**Rationale:** Lighting and appliance usage may respond to common household routines and periods of activity. Consequently, lighting energy may act as a useful indicator of household behavior.

**EDA approach:** Examine the distribution of `lights`, visualize its relationship with `Appliances`, and assess whether the association persists after accounting for environmental and temporal variables.

**Evaluation approach:** Include `lights` in an exploratory model and examine its coefficient, uncertainty, and contribution to predictive performance.

Before retaining this variable in the final deployed model, its availability at the intended prediction time will be verified. A model that uses lighting measurements recorded during the target interval will be interpreted as a contemporaneous estimation model rather than an advance forecasting model.

### Hypothesis 5: Multicollinearity Among Environmental Predictors

**Research hypothesis:** Some indoor sensor and outdoor weather variables are sufficiently correlated to affect the stability and interpretation of Multiple Linear Regression coefficients.

**Null hypothesis (\(H_0\)):** The selected predictors do not exhibit problematic multicollinearity.

**Alternative hypothesis (\(H_1\)):** At least some selected predictors exhibit problematic multicollinearity.

**Rationale:** Temperatures and humidity measurements from different rooms may respond to common environmental conditions. Outdoor temperature, relative humidity, and dew-point temperature may also share overlapping information.

**EDA approach:** Construct a correlation matrix for the predictor variables and identify groups of highly correlated measurements. Calculate variance inflation factors (VIF) after defining an appropriate predictor set.

**Evaluation approach:** Compare coefficient stability across reasonable feature subsets and assess whether multicollinearity affects model interpretation. Where appropriate, consider removing redundant predictors or comparing ordinary least squares with regularized regression as a secondary experiment.

Feature scaling may improve numerical conditioning in some workflows, but it does not eliminate multicollinearity. Predictor selection will therefore be guided by both statistical diagnostics and predictive performance.

## 7. Methodological Considerations and Anticipated Limitations

### 7.1 Temporal Dependence

The observations are recorded sequentially at 10-minute intervals. Nearby observations may be correlated, and household energy consumption may exhibit recurring patterns over time.

A random train/test split could distribute highly similar neighboring observations across both sets, producing an evaluation that does not reflect the difficulty of predicting later observations.

The primary evaluation will therefore use a chronological split, with earlier observations used for training and later observations reserved for testing. Any cross-validation used for model selection will respect temporal ordering where appropriate.

### 7.2 Target Distribution and Unusual Observations

Appliance energy consumption may be skewed, and some observations may have substantially higher values than others.

Because ordinary least squares minimizes the sum of squared residuals, large errors associated with high-consumption observations can exert considerable influence on fitted coefficients.

The target distribution, unusual observations, and potential outliers will be investigated during EDA. Observations will not be removed merely because their values are extreme; any exclusion must be justified by evidence of data quality problems or a clearly defined analytical requirement.

### 7.3 Linearity and Residual Diagnostics

Multiple Linear Regression assumes that the conditional mean of the target can be represented adequately by a linear combination of the specified predictors.

The analysis will examine residual-versus-fitted plots, residual distributions, changing residual variance, and possible temporal structure in the residuals.

A high \(R^2\) alone will not be considered sufficient evidence that the model is appropriate. Model performance will be interpreted alongside residual diagnostics and evaluation on unseen observations.

### 7.4 Multicollinearity and Coefficient Interpretation

Correlated predictors can make individual regression coefficients unstable and difficult to interpret. This issue is particularly relevant to the multiple temperature and humidity measurements in the dataset.

The project will use correlation analysis and VIF diagnostics to investigate potential multicollinearity. The analysis will distinguish between the usefulness of a group of predictors for prediction and the ability to estimate their individual conditional associations reliably.

### 7.5 Generalizability

The dataset represents one low-energy residential building observed over a limited period. The model may learn patterns specific to that building, its occupants, the measurement system, or the environmental conditions during collection.

Consequently, successful evaluation on a held-out period from the same dataset will demonstrate temporal generalization within the available observations, not universal generalization to other households.

External validation on data from additional buildings would be necessary to establish broader applicability.

### 7.6 Feature Availability and Deployment

The deployed model must use features that can genuinely be supplied at prediction time.

Variables measured contemporaneously with the target may be appropriate for estimation but unavailable for advance forecasting. The raw timestamp, lighting measurements, and any future-derived information will therefore be assessed carefully before finalizing the feature set.

The deployed application will use the same preprocessing and feature transformations fitted during model development. This will help ensure that training-time and prediction-time transformations remain consistent.

## 8. Proposed Research Methodology

The project will follow a structured, reproducible machine learning workflow.

1. **Data validation:** Verify the dataset dimensions, data types, duplicate timestamps, missing values, and target measurements.

2. **Exploratory data analysis:** Examine target distribution, temporal trends, predictor distributions, correlations, and unusual observations.

3. **Hypothesis assessment:** Investigate the proposed relationships between indoor environmental conditions, outdoor weather, temporal features, lighting energy, and appliance consumption.

4. **Feature selection:** Define an appropriate set of predictors, exclude the random controls from the primary model, and verify that selected features are available for the intended prediction task.

5. **Baseline modeling:** Establish a simple baseline using a mean-prediction estimator and compare it with an initial Linear Regression model.

6. **Multiple Linear Regression:** Fit a model using selected environmental and temporal predictors, with all preprocessing learned from training data.

7. **Model diagnostics:** Assess linearity, residual behavior, heteroscedasticity, influential observations, and multicollinearity.

8. **Evaluation:** Compare MAE, RMSE, and \(R^2\) on a chronological holdout. Additional time-aware validation may be used during model selection.

9. **Model refinement:** Investigate alternative feature sets and, if justified, compare ordinary least squares with regularized regression approaches such as Ridge Regression.

10. **Deployment:** Save the fitted model and preprocessing pipeline, create a Streamlit prediction interface, and deploy the application using a suitable free hosting tier.

The final model will be selected based on defensible evaluation results, appropriate diagnostics, interpretability, and the intended prediction scenario rather than training performance alone.
