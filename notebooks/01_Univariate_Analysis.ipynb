1. Outdoor temperature varies substantially more than most indoor temperatures.

This makes outdoor temperature a promising candidate for capturing changing environmental conditions that may affect indoor temperature and energy demand.

2. Outdoor humidity features have markedly different distributions.

RH_out is concentrated toward high humidity, with a median of 83.67% and a maximum of 100%. By contrast, RH_6 has exceptionally broad variability, with a standard deviation of 31.15 percentage points and concentrations near very low, moderate, and very high values.

RH_6 deserves particular attention in subsequent analysis because its distribution is much more dispersed than those of most other humidity variables. These observations do not establish that the measurements are erroneous.

3. Wind speed is concentrated at lower values.

The mean is 4.04, the median is 3.67, and the maximum is 14. The prominent concentrations around values 1–4 suggest that lower wind speeds occur frequently, while stronger winds occupy the upper tail.

The median and 75th percentile are both 40, while the histogram has a sharp peak around 40–42. This concentration, together with secondary bumps, suggests that a single symmetric distribution may not describe the variable well.

Press_mm_hg has a relatively small standard deviation of 7.40 but several distributional peaks. Tdewpoint spans −6.6°C to 15.5°C and also shows multiple concentrations.

### 6. Cross-feature patterns discovered
The following are the strongest findings from the univariate analysis as a whole.

1. Many variables are not normally distributed

Appliance consumption is strongly right-skewed, while several temperature and humidity features show multimodality. A normal distribution should not be assumed for all variables.




2. Some features are intermittent, while others are continuous

lights is dominated by zeros, whereas temperature, humidity, pressure, and weather measurements vary across continuous ranges. The different data types and distributions should be considered when evaluating models.




3. Multiple peaks are common

Several temperature and humidity features, as well as pressure, visibility, and dew point, exhibit multiple peaks. These may reflect changing conditions or temporal structure, but their causes cannot be determined from histograms alone.




4. Environmental features may overlap in information

Indoor temperatures, outdoor temperature, dew point, and humidity are physically related. They may each look informative in isolation but contribute redundant information when used together.




5. Extreme values need context, not automatic removal

The high appliance-consumption tail, high RH_5 values, very low RH_6 readings, and the extremes of several temperature features deserve review. Extreme values can be valid observations, and removing them without evidence could discard important forecasting cases.

### 8. Important limitations of the current analysis
The univariate analysis provides a foundation, but it cannot answer several important modeling questions.
- Feature–target relationships: We have not yet established whether higher or lower values of any feature correspond to higher appliance consumption.
- Correlations and redundancy: We have not quantified overlap between temperature, humidity, or weather measurements.
- Temporal patterns: Multimodal distributions may be associated with time, but histograms alone cannot identify daily, weekly, or seasonal patterns.
- Outlier validity: Distributional extremes require contextual assessment before being classified as errors.
- Predictive importance: A feature's spread, skewness, or number of peaks does not determine how much it improves forecast accuracy.
- Forecasting leakage: Features must be checked against the exact prediction timestamp, especially for a 10-minute-ahead target.
One additional consideration is that the observations are time-indexed. A random train/test split can overestimate real-world forecasting performance when nearby observations are distributed across both sets. A chronological evaluation is more appropriate for assessing performance on future periods.

# Univariate EDA Summary — Appliance Energy Forecasting

The univariate exploratory data analysis of the UCI Appliances Energy Prediction dataset examined appliance energy consumption, lighting consumption, indoor temperature and humidity measurements, and outdoor weather variables. The analyzed features contain 19,735 observations each.

The target variable, `Appliances`, is strongly right-skewed and heavy-tailed. Its mean is approximately 97.69 Wh, compared with a median of 60 Wh, and its standard deviation is approximately 102.52 Wh. Most observations are concentrated at relatively low consumption levels, while a smaller number of high-consumption observations extend the upper tail to 1,080 Wh. This distribution suggests that forecasting performance should be assessed across both typical and high-consumption periods, and extreme values should not be removed without contextual evidence.

The `lights` feature is highly sparse, with approximately 77.28% zero observations and a mean of 3.80 Wh. Its intermittent distribution distinguishes it from the continuous environmental measurements and makes it a potentially useful candidate for predicting future appliance consumption, provided that the feature is available at forecast time.

Indoor temperatures generally exhibit relatively narrow distributions, with standard deviations around 1.6–2.2°C for most analyzed rooms. Several temperature variables, including `T3`, `T5`, `T7`, and `T9`, show multiple peaks or secondary clusters. Indoor humidity variables also display varied distributional shapes. Most have moderate variability, while `RH_5` stands out for its greater spread, strong positive skew, and heavier tail. Several other humidity variables show multiple peaks, indicating that their distributions are not well described by a single symmetric concentration.

Outdoor conditions show broader variability in several respects. `T6` and `T_out` cover wide temperature ranges, while `RH_6` has exceptionally high variability and concentrations near very low, moderate, and very high humidity values. `RH_out` is concentrated toward high relative humidity and is left-skewed. Wind speed is concentrated around lower values, with a right tail toward stronger winds. Visibility, atmospheric pressure, and dew point exhibit multimodal distributions with distinct peaks or shoulders.

Across the analyzed features, the principal findings are the strong right skew and heavy tail of appliance consumption, the sparse distribution of lighting usage, the multimodality of several temperature and humidity variables, and the substantial variation in some outdoor weather measurements. These patterns suggest that environmental conditions and lighting consumption are worthwhile candidates for further predictive investigation. However, univariate analysis does not establish causal relationships, correlations, or feature importance.

The next stage of analysis should evaluate feature–target relationships, temporal behavior, feature redundancy, and predictive performance using a forecasting-appropriate evaluation design. Particular care is needed to prevent data leakage by ensuring that every predictor is available at the forecast origin. The ultimate goal is to determine which features improve genuine 10-minute-ahead appliance-energy predictions rather than selecting features solely because their univariate distributions appear interesting.
