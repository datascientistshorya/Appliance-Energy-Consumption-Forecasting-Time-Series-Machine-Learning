# Residential Appliance Energy Forecasting

## Objective
Forecast appliance energy consumption for the next 10-minute
interval using only information available before that interval.

## Dataset
UCI Appliances Energy Prediction dataset.

## Evaluation
Chronological train, validation, and test periods.

## Principles
- Prevent temporal data leakage.
- Establish a baseline before fitting regression models.
- Evaluate using MAE, RMSE, and R-squared.
- Document assumptions, limitations, and reproducibility.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

url = (
    "https://archive.ics.uci.edu/"
    "ml/machine-learning-databases/00374/"
    "energydata_complete.csv"
)

df = pd.read_csv(url)

df["date"] = pd.to_datetime(df["date"], errors="raise")
df = df.sort_values("date").reset_index(drop=True)

print("Dataset loaded successfully")
print("Shape:", df.shape)
print("First timestamp:", df["date"].min())
print("Last timestamp:", df["date"].max())

display(df.head())
```

```python

# 1. Check column data types
print("DATA TYPES")
display(df.dtypes.to_frame(name="data_type"))

# 2. Check missing values
print("\nMISSING VALUES")
missing = df.isnull().sum().sort_values(ascending=False)
display(missing[missing > 0])

print("Total missing cells:", df.isnull().sum().sum())

# 3. Check duplicate records
print("\nDUPLICATE RECORDS")
print("Exact duplicate rows:", df.duplicated().sum())
print("Duplicate timestamps:", df["date"].duplicated().sum())

# 4. Check time intervals
print("\nTIME INTERVALS")
intervals = df["date"].diff().value_counts()
display(intervals.head(10))

# 5. Check target variable
print("\nTARGET SUMMARY")
display(df["Appliances"].describe())

print("Unique target values:", df["Appliances"].nunique())
```
