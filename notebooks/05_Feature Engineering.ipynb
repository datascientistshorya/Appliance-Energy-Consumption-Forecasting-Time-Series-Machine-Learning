### creating lag features

```python
lag_periods = {
    'lag_10min': 1,
    'lag_20min': 2,
    'lag_30min': 3,
    'lag_60min': 6,
    'lag_120min': 12
}

for feature, periods in lag_periods.items():
    df[feature] = df['Appliances'].shift(periods)
```

### Rolling Features

```python
windows = {
    '30min': 3,
    '60min': 6,
    '120min': 12
}

for name, window in windows.items():

    historical = df['Appliances'].shift(1)

    df[f'rolling_mean_{name}'] = (
        historical.rolling(window).mean()
    )

    df[f'rolling_std_{name}'] = (
        historical.rolling(window).std()
    )

    df[f'rolling_max_{name}'] = (
        historical.rolling(window).max()
    )

    df[f'rolling_min_{name}'] = (
        historical.rolling(window).min()
    )
```

### Cyclical Features

```python
# Hour of day
df['hour_sin'] = np.sin(2 * np.pi * df['hour'] / 24)
df['hour_cos'] = np.cos(2 * np.pi * df['hour'] / 24)

# Day of week
df['dayofweek_sin'] = np.sin(
    2 * np.pi * df['day_of_week'] / 7
)
df['dayofweek_cos'] = np.cos(
    2 * np.pi * df['day_of_week'] / 7
)

# Month
df['month_sin'] = np.sin(
    2 * np.pi * (df['month'] - 1) / 12
)
df['month_cos'] = np.cos(
    2 * np.pi * (df['month'] - 1) / 12
)
```

# Feature Engineering — Appliance Energy Forecasting

## 1. Overview

Feature engineering was performed to prepare the UCI Appliances Energy Prediction dataset for the task of forecasting appliance energy consumption at the next 10-minute interval.

The original dataset contains 19,735 observations and 29 columns. The objective of this stage was to create meaningful temporal and environmental features that could help machine learning models identify historical consumption patterns, short-term trends, time-based patterns, and indoor/outdoor environmental relationships.

The feature-engineering process was performed while preserving the chronological structure of the dataset and ensuring that historical features use only information available at the prediction time.

## 2. Target Variable Creation

Created the forecasting target:

- `target_next_10min`: Represents appliance energy consumption at the next 10-minute interval.
- `target_timestamp`: Stores the timestamp corresponding to the target observation.

Both columns were created using `shift(-1)`.

The final observation has a missing target because there is no subsequent observation available in the dataset.

## 3. Lag Feature Engineering

Created five lag features using the historical values of `Appliances`.

| Feature | Purpose |
|---|---|
| `lag_10min` | Appliance consumption 10 minutes earlier |
| `lag_20min` | Appliance consumption 20 minutes earlier |
| `lag_30min` | Appliance consumption 30 minutes earlier |
| `lag_60min` | Appliance consumption 60 minutes earlier |
| `lag_120min` | Appliance consumption 120 minutes earlier |

These features help models identify temporal dependencies and understand how recent and past energy consumption relates to future consumption.

## 4. Rolling Window Feature Engineering

Created rolling statistical features using historical appliance consumption. The rolling calculations were performed after applying `shift(1)` to ensure that the current observation was excluded.

Three rolling windows were considered:

- 30 minutes (3 observations)
- 60 minutes (6 observations)
- 120 minutes (12 observations)

For each window, four statistical features were created:

| Statistic | Purpose |
|---|---|
| Rolling mean | Captures average historical consumption |
| Rolling standard deviation | Captures consumption variability |
| Rolling maximum | Captures peak historical consumption |
| Rolling minimum | Captures minimum historical consumption |

This resulted in 12 rolling features.

These features help models capture short-term consumption trends, fluctuations, and historical consumption ranges.

## 5. Calendar and Cyclical Features

Existing calendar features were retained:

- `hour`
- `minute`
- `day_of_week`
- `day_name`
- `is_weekend`
- `month`
- `day`
- `date_only`
- `day_period`

Additionally, six cyclical features were created:

| Feature | Purpose |
|---|---|
| `hour_sin` | Represents the cyclical position of the hour |
| `hour_cos` | Complements the hour cycle representation |
| `dayofweek_sin` | Represents the cyclical position of the weekday |
| `dayofweek_cos` | Complements the weekday cycle representation |
| `month_sin` | Represents the cyclical position of the month |
| `month_cos` | Complements the month cycle representation |

Sine and cosine transformations preserve the circular nature of time-based variables.

For example, 23:00 and 00:00 are temporally close, even though their numerical hour values appear far apart. Cyclical encoding helps machine learning models represent this relationship.

## 6. Environmental Feature Engineering

Environmental features were created using indoor temperature, indoor humidity, and outdoor weather measurements.

### A. Indoor temperature features

The original indoor temperature measurements (`T1`–`T9`) were used to create:

| Feature | Purpose |
|---|---|
| `indoor_temp_mean` | Represents average indoor temperature |
| `indoor_temp_std` | Captures temperature variability across indoor measurement locations |
| `indoor_temp_max` | Captures the highest indoor temperature |
| `indoor_temp_min` | Captures the lowest indoor temperature |
| `indoor_temp_range` | Represents the difference between maximum and minimum indoor temperatures |

These features summarize indoor temperature conditions and differences between measurement locations.

### B. Indoor–outdoor temperature relationship

- `indoor_outdoor_temp_diff`: Represents the difference between average indoor temperature and outdoor temperature.

This feature captures the thermal difference between indoor and outdoor environments.

### C. Indoor humidity features

The original relative humidity measurements (`RH_1`–`RH_9`) were used to create:

| Feature | Purpose |
|---|---|
| `indoor_humidity_mean` | Represents average indoor relative humidity |
| `indoor_humidity_std` | Captures humidity variability across indoor measurement locations |

These features summarize indoor humidity conditions.

### D. Outdoor environmental feature

Created:

- `outdoor_dewpoint_spread`: Difference between outdoor temperature (`T_out`) and dew point (`Tdewpoint`).

This feature provides information about atmospheric moisture conditions. A smaller spread generally indicates air closer to saturation.

The original outdoor measurements were also retained:

- `T_out`
- `RH_out`
- `Press_mm_hg`
- `Windspeed`
- `Visibility`
- `Tdewpoint`

## 8. Missing Values and Dataset Validation

After feature engineering, the dataset contained:

- Rows: 19,735
- Columns: 78

The missing values were limited to expected locations:

- One missing target observation at the end of the dataset.
- Missing lag values at the beginning due to insufficient historical observations.
- Missing rolling statistics at the beginning due to incomplete rolling windows.

No unexpected missing values were observed in the newly engineered environmental features.

The missing values will be handled during final data preparation rather than being filled arbitrarily with zero.

The chronological structure of the dataset was preserved throughout feature engineering.

## 9. Key Considerations

Several considerations will guide the next stages:

1. The forecasting target represents appliance consumption at the next 10-minute interval.
2. Lag and rolling features use historical appliance consumption.
3. Target-related columns must not be included as model predictors.
4. All candidate features must be checked for availability at the prediction origin.
5. Feature usefulness and redundancy will be evaluated during modeling.
6. Chronological train, validation, and test splits will be used instead of random splitting.

Feature creation does not automatically guarantee improved model performance. The usefulness of the engineered features will ultimately be assessed through time-based model evaluation.
